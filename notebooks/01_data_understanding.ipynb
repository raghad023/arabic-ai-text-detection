```python
from itertools import combinations

subset_names = list(dataset.keys())

original_abstract_sets = {
    subset_name: set(dataset[subset_name]["original_abstract"])
    for subset_name in subset_names
}

for subset_a, subset_b in combinations(subset_names, 2):
    overlap = original_abstract_sets[subset_a] & original_abstract_sets[subset_b]

    print(
        f"{subset_a} vs {subset_b}: "
        f"{len(overlap)} overlapping original abstracts"
    )

common_to_all = set.intersection(*original_abstract_sets.values())

print(
    f"\nCommon to all three subsets: "
    f"{len(common_to_all)} original abstracts"
)
```

```python
human_texts = set()

ai_texts = set()

ai_columns = [
    "allam_generated_abstract",
    "jais_generated_abstract",
    "llama_generated_abstract",
    "openai_generated_abstract"
]

for subset_name in dataset.keys():
    human_texts.update(
        dataset[subset_name]["original_abstract"]
    )

    for column in ai_columns:
        ai_texts.update(
            dataset[subset_name][column]
        )

human_ai_overlap = human_texts & ai_texts

print(f"Unique human texts: {len(human_texts)}")
print(f"Unique AI-generated texts: {len(ai_texts)}")
print(
    f"Exact Human-AI text overlaps: "
    f"{len(human_ai_overlap)}"
)
```

```python
overlap_records = []

for subset_name in dataset.keys():
    df = dataset[subset_name].to_pandas()

    for row_index, row in df.iterrows():

        for column in ai_columns:

            ai_text = row[column]

            if ai_text in human_ai_overlap:

                overlap_records.append({
                    "subset": subset_name,
                    "row_index": row_index,
                    "generator": column,
                    "same_as_own_original":
                        ai_text == row["original_abstract"],
                    "text": ai_text
                })

overlap_df = pd.DataFrame(overlap_records)

overlap_df
```

```python
for subset_name in dataset.keys():
    df = dataset[subset_name].to_pandas()

    print(f"\nLength outliers in {subset_name}:")

    for column in df.columns:
        lengths = df[column].str.len()

        q1 = lengths.quantile(0.25)
        q3 = lengths.quantile(0.75)
        iqr = q3 - q1

        lower_bound = q1 - 1.5 * iqr
        upper_bound = q3 + 1.5 * iqr

        outlier_count = (
            (lengths < lower_bound) |
            (lengths > upper_bound)
        ).sum()

        print(
            f"{column}: "
            f"{outlier_count} length outliers "
            f"(upper bound = {upper_bound:.2f})"
        )
```

```python
all_raw_texts = []

for subset_name in dataset.keys():
    for column in dataset[subset_name].column_names:
        all_raw_texts.extend(
            dataset[subset_name][column]
        )


characters_to_check = {
    "Plain Alef (ا)": "ا",
    "Alef with Hamza Above (أ)": "أ",
    "Alef with Hamza Below (إ)": "إ",
    "Alef with Madda (آ)": "آ",
    "Alef Wasla (ٱ)": "ٱ",
    "Yaa (ي)": "ي",
    "Alef Maqsura (ى)": "ى",
    "Hamza (ء)": "ء",
    "Waw with Hamza (ؤ)": "ؤ",
    "Yaa with Hamza (ئ)": "ئ",
    "Tatweel (ـ)": "ـ",
}

character_stats = []

for name, character in characters_to_check.items():
    total_occurrences = sum(
        text.count(character)
        for text in all_raw_texts
    )

    texts_containing = sum(
        character in text
        for text in all_raw_texts
    )

    character_stats.append({
        "character": name,
        "total_occurrences": total_occurrences,
        "texts_containing_character": texts_containing
    })

character_stats_df = pd.DataFrame(character_stats)

character_stats_df
```

```python
arabic_diacritics = {
    "\u064B",  # Fathatan
    "\u064C",  # Dammatan
    "\u064D",  # Kasratan
    "\u064E",  # Fatha
    "\u064F",  # Damma
    "\u0650",  # Kasra
    "\u0651",  # Shadda
    "\u0652",  # Sukun
    "\u0670",  # Dagger Alef
}

diacritic_occurrences = sum(
    sum(char in arabic_diacritics for char in text)
    for text in all_raw_texts
)

texts_with_diacritics = sum(
    any(char in arabic_diacritics for char in text)
    for text in all_raw_texts
)

print(f"Total diacritic occurrences: {diacritic_occurrences}")
print(f"Texts containing diacritics: {texts_with_diacritics}")
```

```python
import unicodedata

unicode_issues = {
    "Replacement character (�)": "\uFFFD",
    "Zero Width Space": "\u200B",
    "Zero Width Non-Joiner": "\u200C",
    "Zero Width Joiner": "\u200D",
    "Left-to-Right Mark": "\u200E",
    "Right-to-Left Mark": "\u200F",
    "Byte Order Mark": "\uFEFF",
}

unicode_issue_stats = []

for name, character in unicode_issues.items():
    occurrences = sum(
        text.count(character)
        for text in all_raw_texts
    )

    texts_containing = sum(
        character in text
        for text in all_raw_texts
    )

    unicode_issue_stats.append({
        "character": name,
        "total_occurrences": occurrences,
        "texts_containing_character": texts_containing
    })

unicode_issue_stats_df = pd.DataFrame(unicode_issue_stats)

unicode_issue_stats_df
```

```python
unicode_issue_records = []

for subset_name in dataset.keys():
    df = dataset[subset_name].to_pandas()

    for column in df.columns:
        for row_index, text in df[column].items():

            found_issues = [
                name
                for name, character in unicode_issues.items()
                if character in text
            ]

            if found_issues:
                unicode_issue_records.append({
                    "subset": subset_name,
                    "row_index": row_index,
                    "column": column,
                    "issues": ", ".join(found_issues),
                    "text_preview": repr(text[:250])
                })

unicode_issue_df = pd.DataFrame(unicode_issue_records)

unicode_issue_df
```

```python
from collections import Counter
import unicodedata

hidden_unicode_counts = Counter()
hidden_unicode_text_counts = Counter()

for text in all_raw_texts:
    characters_found_in_text = set()

    for char in text:
        category = unicodedata.category(char)

        if category in {"Cf", "Cc"} and char not in {"\n", "\r", "\t"}:
            hidden_unicode_counts[char] += 1
            characters_found_in_text.add(char)

    for char in characters_found_in_text:
        hidden_unicode_text_counts[char] += 1


hidden_unicode_records = []

for char, count in hidden_unicode_counts.items():
    hidden_unicode_records.append({
        "character": repr(char),
        "code_point": f"U+{ord(char):04X}",
        "unicode_name": unicodedata.name(char, "UNKNOWN"),
        "category": unicodedata.category(char),
        "total_occurrences": count,
        "texts_containing_character": hidden_unicode_text_counts[char]
    })

hidden_unicode_df = pd.DataFrame(hidden_unicode_records)

hidden_unicode_df
```

```python
target_char = "\u202A"

for subset_name in dataset.keys():
    df = dataset[subset_name].to_pandas()

    for column in df.columns:
        matches = df[
            df[column].str.contains(target_char, regex=False)
        ]

        if not matches.empty:
            for row_index, row in matches.iterrows():
                print(f"Subset: {subset_name}")
                print(f"Row index: {row_index}")
                print(f"Column: {column}")
                print(f"Text preview: {repr(row[column][:300])}")
```

```python
def is_arabic_letter(char):
    return (
        char.isalpha()
        and "ARABIC" in unicodedata.name(char, "")
    )


def is_latin_letter(char):
    return (
        char.isalpha()
        and "LATIN" in unicodedata.name(char, "")
    )


for subset_name in dataset.keys():
    df = dataset[subset_name].to_pandas()

    print(f"\nLanguage/script checks in {subset_name}:")

    for column in df.columns:
        no_arabic_count = 0
        mixed_arabic_latin_count = 0

        for text in df[column]:
            has_arabic = any(
                is_arabic_letter(char)
                for char in text
            )

            has_latin = any(
                is_latin_letter(char)
                for char in text
            )

            if not has_arabic:
                no_arabic_count += 1

            if has_arabic and has_latin:
                mixed_arabic_latin_count += 1

        print(
            f"{column}: "
            f"no Arabic letters = {no_arabic_count}, "
            f"Arabic+Latin mixed = {mixed_arabic_latin_count}"
        )
```

```python
artifact_patterns = {
    "بالطبع": r"بالطبع",
    "إليك": r"إليك",
    "ملخص:": r"ملخص\s*:",
    "العنوان:": r"العنوان\s*:",
    "الكلمات المفتاحية:": r"الكلمات\s+المفتاحية\s*:",
    "المؤلفون:": r"المؤلفون\s*:",
    "Generated Abstract": r"generated\s+abstract",
    "Abstract:": r"\babstract\s*:",
    "Title:": r"\btitle\s*:",
}

artifact_records = []

for subset_name in dataset.keys():
    df = dataset[subset_name].to_pandas()

    for column in df.columns:
        for artifact_name, pattern in artifact_patterns.items():

            count = (
                df[column]
                .str.contains(
                    pattern,
                    case=False,
                    regex=True,
                    na=False
                )
                .sum()
            )

            if count > 0:
                artifact_records.append({
                    "subset": subset_name,
                    "column": column,
                    "artifact": artifact_name,
                    "count": count
                })

artifact_df = pd.DataFrame(artifact_records)

artifact_df
```
